```python
    # Які категорії товарів дають найбільший обсяг продажів і валовий дохід?

product_analysis = df.groupby('Product line').agg({'Total':'sum','Quantity':'sum','gross income':'sum'})

product_analysis.sort_values('Total', ascending=False)
```

*Створюємо функцію **purchase_size**, яка визначає категорію розміру покупки за значенням Quantity: 1-3 товари -Small, 4 -7 -Medium, 8 -10 -Large.*

```python
def purchase_size(quantity):     # Чи змінюється задоволеність клієнта залежно від розміру покупки?
    if quantity <= 3:
        return 'Small'
    elif quantity <= 7:
        return 'Medium'
    else:
        return 'Large'
```

*Створюємо нову колонку **Purchase size**, у яку за допомогою apply() записуємо результат роботи функції purchase_size для кожного значення колонки Quantity.*

```python
df['Purchase size'] = df['Quantity'].apply(purchase_size)
```

```python
df[['Quantity', 'Rating', 'Purchase size']]
```

*Групуємо покупки за **Purchase size** та знаходимо середній **Rating** і середній **Total** для кожної групи*.

```python
purchase_analysis = df.groupby('Purchase size').agg({'Rating':'mean','Total':'mean'})

purchase_analysis
```
