### ЗАВДАННЯ 1. Розрахунок розміру вибірки

> Для початку, уявімо, що ми тільки плануємо проведення зазначеного A/B-тесту і хочемо зрозуміти, дані про скількох користувачів нам треба зібрати, аби досягнути відчутного ефекту.
>
> Відчутним ефектом ми вважатимемо збільшення утримання на 1% після внесення зміни.
>
> Обчисліть, скільки користувачів сумарно нам треба аби досягнути такого ефекту, якщо продакт менеджер нам повідомив, що базове утримання є 19%.

---



###  Розрахунок розміру вибірки

**Дано:**
- Базове утримання: 19% (p₁ = 0.19).
- Очікуване утримання: 20% (p₂ = 0.20).
- Мінімальний ефект: +1 відсотковий пункт.
- Рівень значущості: α = 0.05.
- Статистична потужність: 80%.
- Розподіл користувачів між групами: 50/50.

Для визначення необхідної кількості користувачів використовуємо `NormalIndPower()` із бібліотеки `statsmodels`.


````python

import numpy as np
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

p1 = 0.19
p2 = 0.20
alpha = 0.05
power = 0.80

effect_size = proportion_effectsize(p1, p2)

n = NormalIndPower().solve_power(
    effect_size=effect_size,
    alpha=alpha,
    power=power,
    ratio=1,
    alternative="two-sided"
)

n_group = int(np.ceil(n))

print(f"Користувачів на одну групу: {n_group}")
print(f"Загальна кількість користувачів: {n_group * 2}")
````


### Висновок

Для виявлення збільшення утримання гравців із **19% до 20%** необхідно залучити щонайменше **49 276 користувачів**:

- Контрольна група (gate_30): 24 638.
- Тестова група (gate_40): 24 638.

Такий розмір вибірки забезпечує статистичну потужність 80% при рівні значущості 5% для виявлення заданого ефекту.

### Аналіз утримання на 7-й день

Завантажимо дані A/B-тесту та порівняємо середнє значення показника `retention_7` для контрольної групи (`gate_30`) і тестової групи (`gate_40`).


````python

import pandas as pd

df = pd.read_csv("cookie_cats.csv")

print("Середнє утримання на 7-й день:")
print(df.groupby("version")["retention_7"].mean())
````


### Формулювання гіпотез

**H₀:** p₃₀ = p₄₀ — утримання гравців на 7-й день однакове для обох версій гри.

**H₁:** p₃₀ ≠ p₄₀ — утримання гравців на 7-й день відрізняється між версіями гри.

**Тип тесту:** двосторонній Z-тест для двох незалежних пропорцій.

Мета — перевірити, чи вплинуло перенесення воріт із 30-го на 40-й рівень на утримання гравців через 7 днів після встановлення гри.



### Висновок

Середнє утримання гравців на 7-й день становить:

- **gate_30 — 19.02%**
- **gate_40 — 18.20%**

Утримання в контрольній групі вище на **0.82 відсоткового пункту**.

Попередній висновок: версія **gate_30** демонструє краще утримання гравців через 7 днів після встановлення гри. Статистичну значущість різниці необхідно перевірити окремо.

### ЗАВДАННЯ 3. Z-тест та довірчі інтервали

> Перевірте з допомогою пасуючого варіанту z-тесту, чи дає якась з версій гри кращий показник `retention_7` на рівні значущості 0.05.

> де замість `...` — обчислені значення.

> 2. Чи перетинаються довірчі інтервали утримання користувачів з різних версій гри? Про що це каже?

### Z-тест та довірчі інтервали

**Гіпотези:**

H₀: p₃₀ = p₄₀ — утримання гравців однакове для обох версій.

H₁: p₃₀ ≠ p₄₀ — утримання гравців відрізняється між версіями.

**Тип тесту:** двосторонній Z-тест для двох незалежних пропорцій.

Рівень значущості: α = 0.05.

Додатково обчислюємо 95% довірчі інтервали для обох груп.


````python

from statsmodels.stats.proportion import proportions_ztest, proportion_confint

# Дані по групах
groups = df.groupby("version")["retention_7"].agg(["sum", "count"])

successes = groups.loc[["gate_30", "gate_40"], "sum"]
nobs = groups.loc[["gate_30", "gate_40"], "count"]

# Z-тест
z_stat, p_value = proportions_ztest(successes, nobs, alternative="two-sided")

# Довірчі інтервали
ci_control = proportion_confint(successes["gate_30"], nobs["gate_30"], alpha=0.05)
ci_treatment = proportion_confint(successes["gate_40"], nobs["gate_40"], alpha=0.05)

print(f"z statistic: {z_stat:.4f}")
print(f"p-value: {p_value:.5f}")
print(f"Довірчий інтервал 95% для групи control: [{ci_control[0]:.4f}, {ci_control[1]:.4f}]")
print(f"Довірчий інтервал 95% для групи treatment: [{ci_treatment[0]:.4f}, {ci_treatment[1]:.4f}]")
````


### Висновок

**1. Чи є статистично значуща різниця?**

Так. Значення p-value = 0.00155 < 0.05, тому відхиляємо H₀. Різниця між групами є статистично значущою.

Утримання на 7-й день у групі gate_30 (19.02%) вище, ніж у групі gate_40 (18.20%), на 0.82 відсоткового пункту.

**2. Чи перетинаються довірчі інтервали?**

Ні. 95% довірчі інтервали не перетинаються:

- control (gate_30): [18.66%; 19.38%]
- treatment (gate_40): [17.85%; 18.55%]

Це додатково підтверджує статистично значущу різницю між групами.

**Загальний висновок:** перенесення воріт із 30-го на 40-й рівень негативно вплинуло на утримання гравців на 7-й день. Версія gate_30 демонструє кращий результат.

### ЗАВДАННЯ 4. Тест Хі-квадрат (χ²)

> Виконайте тест Хі-квадрат на рівні значущості 5%, аби визначити, чи є залежність між версією гри та утриманням гравця на 7-й день після реєстрації.
>
>
> 1. Напишіть, як для цього тесту будуть сформульовані гіпотези.

### χ²-тест незалежності

**H₀:** Версія гри та утримання гравців на 7-й день незалежні між собою.

**H₁:** Існує статистично значущий зв'язок між версією гри та утриманням гравців на 7-й день.

**Тип тесту:** χ²-тест незалежності (Хі-квадрат).

**Рівень значущості:** α = 0.05 (5%).


````python

from scipy.stats import chi2_contingency

# Таблиця спряженості
table = pd.crosstab(df["version"], df["retention_7"])

# Тест Хі-квадрат
chi2, p_value, dof, expected = chi2_contingency(table)

print(table)
print(f"\nχ²-статистика: {chi2:.4f}")
print(f"Ступені свободи: {dof}")
print(f"p-value: {p_value:.5f}")
````

````python

# Висновок за результатами χ²-тесту
alpha = 0.05

if p_value < alpha:
    print(f"Відхиляємо H₀: версія гри та утримання гравців на 7-й день статистично залежні (p-value = {p_value:.5f} < {alpha})")
else:
    print(f"Не відхиляємо H₀: статистично значущого зв'язку не виявлено (p-value = {p_value:.5f} ≥ {alpha})")
````


### Висновок за результатами χ²-тесту

За результатами χ²-тесту отримали:

- χ²-статистика = 9.9591
- Ступені свободи = 1
- p-value = 0.00160
- Рівень значущості α = 0.05

Значення p-value = 0.00160 < 0.05, тому відхиляємо нульову гіпотезу H₀.

**Висновок:** між версією гри та утриманням гравців на 7-й день існує статистично значущий зв'язок.

Версія gate_30 має вище утримання (19.02%) порівняно з gate_40 (18.20%). Різниця становить 0.82 відсоткового пункту.

Отже, результати χ²-тесту підтверджують, що перенесення воріт із 30-го на 40-й рівень вплинуло на утримання гравців на 7-й день.
