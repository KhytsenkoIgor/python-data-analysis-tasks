## Завдання 3. Задача

E-commerce компанія аналізує **середню тривалість дзвінків у відділі підтримки клієнтів**.
Раніше середня тривалість дзвінка становила **143 секунди**.
Після оптимізації скриптів спілкування менеджер хоче перевірити, чи **зменшився середній час обслуговування клієнта**.

Було зібрано випадкову вибірку з **100 дзвінків**:

- середня тривалість: $\bar{x}$ = 133 секунд,
- стандартне відхилення: $\sigma$ = 35 секунд.

Виконайте перевірку гіпотези на рівні значущості $\alpha = 0.05$ з ручними обчисленнями (поки без Python, просто з калькулятором і таблицею критичних значень, як було в прикладі в лекції).


## Розв’язок задачі.

**1. Гіпотези:**

Позначимо $\mu$ як середню тривалість дзвінків після оптимізації скриптів.

- Нульова гіпотеза: $H_0: \mu = 143$ сек.
- Альтернативна гіпотеза: $H_a: \mu < 143$ сек.

Перевіряємо зменшення середньої тривалості дзвінків, маємо **лівосторонній Z-тест**.

**2. Рівень значущості:**

$\alpha = 0.05$

**3. Критичне значення (односторонній тест) :**

У лівому хвості шукаємо $z_{\alpha} = z_{0.05}$.

За Z-таблицею:

$Z_{кр} \approx -1.645$

**Область відхилення:** $Z \leq -1.645$.

**4. Розрахунок Z-статистики :**

Відомі дані:

- $\bar{x} = 133$ сек. - середня тривалість дзвінків у вибірці.
- $\mu_0 = 143$ сек. - попередня середня тривалість дзвінка.
- $\sigma = 35$ сек. - стандартне відхилення.
- $n = 100$ - кількість дзвінків у вибірці.

Формула одновибіркового Z-тесту:

$Z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$

Підставляємо значення:

$Z = \frac{133 - 143}{35 / \sqrt{100}} = \frac{-10}{35 / 10} = \frac{-10}{3.5} \approx -2.857$


**5. Порівняння та висновок:**

$Z_{розр} = -2.857 < Z_{кр} = -1.645$

Розрахункова Z-статистика потрапляє до області відхилення нульової гіпотези.

Відхиляємо H₀. На рівні значущості 5% твердження про зменшення середньої тривалості дзвінків після оптимізації скриптів порівняно зі 143 сек. підтверджується статистичними даними.
