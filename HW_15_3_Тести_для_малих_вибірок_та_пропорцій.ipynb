# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.
2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.



## Крок 1. Формулюємо гіпотези

**Гіпотези**

H₀: μ = 123.7 (середній час до покупки після редизайну не збільшився)

H₁: μ > 123.7 (середній час до покупки після редизайну збільшився)

Тип тесту: правосторонній (one-tailed).

**Чому саме правосторонній тест?**

Компанія припускає, що після редизайну середній час до покупки **збільшився** (а не просто змінився). Тому ми перевіряємо, чи середній час **більший за 123.7 хвилини**.


## Крок 2. Обчислюємо вибіркові статистики: x̄, s, n

```python

import numpy as np

data = [128, 135, 121, 142, 126, 151, 123]

n = len(data)
x_mean = np.mean(data)
s = np.std(data, ddof=1)

print(f"Розмір вибірки (n): {n}")
print(f"Вибіркове середнє (x̄): {x_mean:.2f} хв")
print(f"Вибіркове стандартне відхилення (s): {s:.2f} хв")
```


## Крок 3. Проводимо t-тест Стьюдента

Оскільки вибірка мала (n = 7), а стандартне відхилення генеральної сукупності невідоме, використовуємо **одновибірковий правосторонній t-тест** при α = 0.10.


```python

from scipy import stats
import numpy as np

mu_0 = 123.7
alpha = 0.10
df = n - 1

# Розрахунок t-статистики
t_stat = (x_mean - mu_0) / (s / np.sqrt(n))

# Критичне значення
t_crit = stats.t.ppf(1 - alpha, df)

# p-value для правостороннього тесту
p_value = stats.t.sf(t_stat, df)

print(f"Ступені свободи (df): {df}")
print(f"t-статистика: {t_stat:.4f}")
print(f"Критичне значення t: {t_crit:.4f}")
print(f"p-value: {p_value:.4f}")
```


**Результати t-тесту**

- Рівень значущості: α = 0.10.
- Ступені свободи: df = 6.
- t-статистика: 2.0687.
- Критичне значення: t = 1.4398.
- p-value = 0.0420.

Оскільки t-статистика (2.0687) перевищує критичне значення (1.4398), результат потрапляє в область відхилення H₀.



### Крок 3. Спрощений розрахунок t-тесту


```python

# scipy тест
res = stats.ttest_1samp(data, popmean=123.7, alternative="greater")
print(f"scipy: t = {res.statistic:.3f}, p-value = {res.pvalue:.5f}")
```

```python
res
```


## Крок 4. Приймаємо рішення на основі p-value

**Правило прийняття рішення:**

- Якщо p-value < α — відхиляємо H₀.
- Якщо p-value ≥ α — не відхиляємо H₀.

Рівень значущості: α = 0.10.


```python
if res.pvalue < alpha:
    print(f"Відхиляємо H₀: середній час до покупки збільшився на рівні значущості {alpha} (p-value = {res.pvalue:.5f} < {alpha})")
else:
    print(f"Не відхиляємо H₀: доказів збільшення часу до покупки недостатньо (p-value = {res.pvalue:.5f} ≥ {alpha})")
```


## Крок 5. Висновок

Вибірковий середній час до покупки після редизайну становить **132.29 хв**, що більше за історичне значення **123.7 хв** на **8.59 хв**.

Оскільки **p-value < 0.10**, відхиляємо нульову гіпотезу H₀.

**Висновок:** на рівні значущості 10% є статистично значущі докази того, що після редизайну сайту середній час до покупки збільшився.


## Крок 6. Перевірка при α = 0.05

```python
alpha = 0.05

if res.pvalue < alpha:
    print(f"Відхиляємо H₀: середній час до покупки збільшився на рівні значущості {alpha} (p-value = {res.pvalue:.5f} < {alpha})")
else:
    print(f"Не відхиляємо H₀: доказів збільшення часу до покупки недостатньо (p-value = {res.pvalue:.5f} ≥ {alpha})")
```


## Висновок при α = 0.05

При зменшенні рівня значущості з 0.10 до 0.05 рішення **не змінюється**.

Оскільки p-value ≈ 0.042 < 0.05, відхиляємо H₀.

**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?



### Крок 1. Формулюємо гіпотези

**Дано:**
- Початкова частка впізнаваності логотипу: p₀ = 0.23 (23%).
- Розмір вибірки: n = 1200.
- Кількість дорослих, які впізнали логотип: x = 311.
- Рівень значущості: α = 0.03 (3%).

**Гіпотези**

H₀: p = 0.23 (частка дорослих, які впізнають логотип, не збільшилася).

H₁: p > 0.23 (частка дорослих, які впізнають логотип, збільшилася).

**Тип тесту:** правосторонній одновибірковий Z-тест для пропорції.

**Чому саме правосторонній тест?**

Компанія хоче перевірити, чи після рекламної кампанії впізнаваність логотипу **зросла**, а не просто змінилася. Тому перевіряємо, чи нова частка перевищує 23%.



### Крок 2. Обчислюємо вибіркову частку та проводимо Z-тест

Для перевірки гіпотези використовуємо **одновибірковий правосторонній Z-тест для пропорції**.

Оскільки вибірка велика (n = 1200), застосовуємо бібліотеку `statsmodels` для розрахунку Z-статистики та p-value.


```python

from statsmodels.stats.proportion import proportions_ztest

# Вхідні дані
x = 311
n = 1200
p0 = 0.23
alpha = 0.03

# Вибіркова частка
p_hat = x / n

# Z-тест
z_stat, p_value = proportions_ztest(x, n, value=p0, alternative="larger")

print(f"Вибіркова частка: {p_hat:.4f} ({p_hat:.2%})")
print(f"Z-статистика: {z_stat:.4f}")
print(f"p-value: {p_value:.5f}")
```


### Крок 3. Прийняття рішення на основі p-value

**Правило прийняття рішення:**

- Якщо p-value < α — відхиляємо H₀.
- Якщо p-value ≥ α — не відхиляємо H₀.

Рівень значущості: α = 0.03.


```python
if p_value < alpha:
    print(f"Відхиляємо H₀: впізнаваність логотипу зросла вище 23% на рівні значущості {alpha} (p-value = {p_value:.5f} < {alpha})")
else:
    print(f"Не відхиляємо H₀: доказів зростання впізнаваності логотипу недостатньо (p-value = {p_value:.5f} ≥ {alpha})")
```


### Крок 4. Висновок

До рекламної кампанії впізнаваність логотипу становила **23%**, а після кампанії — **25.92%**.

Таким чином, впізнаваність зросла на **2.92 відсоткового пункту**.

Оскільки **p-value = 0.01056 < α = 0.03**, відхиляємо нульову гіпотезу H₀.

**Висновок:** на рівні значущості 3% є статистично значущі докази того, що частка дорослих, які впізнають логотип компанії, перевищує 23%.



### Крок 5. Обчислення 97% довірчого інтервалу

Обчислюємо довірчий інтервал для частки дорослих, які впізнають логотип компанії, з довірчою ймовірністю **97%**, так як рівень значущості α = 0.03 (3%).

Для розрахунку використовуємо нормальне наближення (`method="normal"`).


```python

from statsmodels.stats.proportion import proportion_confint

ci_low, ci_high = proportion_confint(x, n, alpha=0.03, method="normal")

print(f"97% довірчий інтервал: [{ci_low:.4f}; {ci_high:.4f}]")
print(f"У відсотках: [{ci_low:.2%}; {ci_high:.2%}]")
```


### Інтерпретація довірчого інтервалу

97% довірчий інтервал для впізнаваності логотипу становить приблизно **23.17%–28.66%**.

Нижня межа інтервалу перевищує початковий рівень **23%**.

**Висновок:** довірчий інтервал додатково підтверджує статистично значуще зростання впізнаваності логотипу. Це узгоджується з результатом Z-тесту, за яким ми відхилили H₀ на рівні значущості 3%.
